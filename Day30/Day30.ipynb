```python
# import modules
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
```

```python
# Data
data = pd.read_csv('customer_data.csv')
df = pd.DataFrame(data)
print(df.head())
df.shape
```

```python
# First 5 features
df = df.iloc[:,:7]
print(df.head())
df.shape
```

```python
# Handle missing values
df = df.dropna()
df.isnull().sum()
```

```python
# Handle duplicates
df = df.drop_duplicates()
df.shape
```

```python
# Data Visualization
plt.figure(figsize=(8,6))
sns.countplot(x=df['fea_1'])
plt.title("Features")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
# Heatmap
corr = df[['fea_1','fea_2','fea_3','fea_4','fea_5']].corr(numeric_only=True)
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap='viridis')
plt.show()
```

```python
# Linear regression
x = df[['fea_1']]
y = df[['fea_3']]

x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2, random_state=42)
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
# Predictions
y_pred = model.predict(x_test)

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
print(f"Mean Squared Error: {mse:.4f}")
print(f"R2 Score: {r2:.4f}")
```

```python
# Regression plot
plt.figure(figsize=(6,4))
sns.regplot(x=x_test, y=y_test, scatter_kws={'color':'b'}, line_kws={'color':'red'})
plt.xlabel("Feature 1")
plt.ylabel("Feature 3")
plt.title("Regression Plot (Test Set)")
plt.show()
```

```python
# Violin plot
plt.figure(figsize=(6,4))
sns.violinplot(x=df['fea_1'])
plt.title("Box Plot for Feature 1")
plt.show()
```
