# Imbalanced Datasets

```python
# import modules
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, mean_absolute_error
from imblearn.over_sampling import RandomOverSampler
```

```python
# Data
x,y = make_classification(n_classes=2,class_sep=2,weights=[0.85,0.15],n_informative=2,n_redundant=0,n_features=2,random_state=42,n_samples=1000)
```

```python
# Visualization
plt.scatter(x[y==0][:,0],x[y==0][:,1],label='Healthy (Class0)',alpha=0.5)
plt.scatter(x[y==1][:,0],x[y==1][:,1],label='Healthy (Class1)',alpha=0.5)
plt.title('Class Distribution')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.show()
```

```python
# Model
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.3, random_state=42)

model = LogisticRegression()
model.fit(x_train,y_train)
```

```python
# Predictions
y_pred = model.predict(x_test)
print(f"Mean Absolute Error: {mean_absolute_error(y_test, y_pred)}")
```

```python
# Handling Imbalanced Dataset
# Random Oversampling

ros = RandomOverSampler(random_state=42)
x_sample, y_sample = ros.fit_resample(x_train,y_train)
```

```python
# Visualization
plt.scatter(x_sample[y_sample==0][:,0],x_sample[y_sample==0][:,1],label='Healthy (Class0)',alpha=0.5)
plt.scatter(x_sample[y_sample==1][:,0],x_sample[y_sample==1][:,1],label='Healthy (Class1)',alpha=0.5)
plt.title('Class Distribution')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.show()
```

```python
# Distributions
print("Original Class Distribution:")
print(f"Class 0: {np.sum(y_train==0)}, \nClass 1: {np.sum(y_train==1)}")

print("ROS Class Distribution:")
print(f"Class 0: {np.sum(y_sample==0)}, \nClass 1: {np.sum(y_sample==1)}")
```

```python
# Random Forest & Grid Search
rf = RandomForestClassifier(random_state=42)
params = {'n_estimators':[50,100,200],
          'max_depth':[None,10,20,30],
          'min_samples_split':[2,5,10]
}

x_train, x_test, y_train, y_test = train_test_split(x_sample,y_sample, test_size=0.3, random_state=42)

grid_search = GridSearchCV(estimator=rf,param_grid=params,scoring='accuracy',verbose=1,cv=5,n_jobs=-1)
grid_search.fit(x_train,y_train)
```

```python
# Best Parameters
best_params = grid_search.best_params_
best_params
```

```python
# Best Model
best_model = grid_search.best_estimator_
```

```python
# Predictions
y_pred = best_model.predict(x_test)
print(f"Mean Absolute Error: {mean_absolute_error(y_test, y_pred)}")
```
