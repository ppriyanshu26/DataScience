```python
# import modules
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from keras.models import Sequential
from keras.layers import Dense, LSTM
from sklearn.metrics import root_mean_squared_error
```

```python
# Data

data = np.array([i for i in range(100)]) #[0,1,2,3,...]
data = data.reshape(-1,1)                #[[0],[1],[2],[3],...]

scaler = MinMaxScaler(feature_range=(0,1))
data_scaled = scaler.fit_transform(data)
```

```python
# Sequences
def create_sequences(data,time_steps=5):
    x,y = [],[]
    for i in range(len(data)-time_steps):
        x.append(data[i:i+time_steps])
        y.append(data[i+time_steps])
    return np.array(x),np.array(y)

time_steps = 5
x,y = create_sequences(data_scaled,time_steps)

x = x.reshape((x.shape[0],x.shape[1],1))
```

```python
# Splitting

split = int(len(x)*0.8)

x_train,x_test = x[:split],x[split:]
y_train,y_test = y[:split],y[split:]
```

```python
# Model

model = Sequential([
    LSTM(50,return_sequences=False,input_shape=(time_steps,1)),
    Dense(1)
])
model.compile(optimizer='adam',loss='mse')
model.fit(x_train,y_train,epochs=20,batch_size=1,verbose=1)
```

```python
# Predictions

y_pred = model.predict(x_test)

plt.figure(figsize=(10,5))
plt.plot(y_test,label='Actual')
plt.plot(y_pred, label='Predicted')
plt.legend()
plt.show()
```

```python
# Accuracy

print("RMSE:",root_mean_squared_error(y_test,y_pred))
```

```python
# Test
test_input = np.array([95,96,97,98,99]).reshape(1,time_steps,1)
test_input_scaled = scaler.transform(test_input.reshape(-1,1)).reshape(1,time_steps,1)

predicted_value_scaled = model.predict(test_input_scaled)
predicted_value = scaler.inverse_transform(predicted_value_scaled.reshape(-1,1))
print(predicted_value[0][0])
```
