# NLP


```python
# import modules
from nltk.tokenize import word_tokenize,sent_tokenize
from nltk.probability import FreqDist
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords
```

```python
# Sample Text
text = "NLP is fun. NLP is amazing."
```

```python
# Tokenization
tokens = sent_tokenize(text.lower())
print("Sentence Tokenize:",tokens)

tokens = word_tokenize(text.lower())
print("Word Tokenize:",tokens)
```

```python
# Calculate frequency
fdist = FreqDist(tokens)

print(fdist)
print(dict(fdist))

for word, freq in fdist.items():
    print(word, ":", freq)
```

```python
# Example words
words = ['running', 'happiness', 'better']
```

```python
# Stemming
stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(i) for i in words]
print('Stemmed:',stemmed_words)
```

```python
# Lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(i, pos='v') for i in words]
print('Lemmatized:',lemmatized_words)
```

```python
# Sample text
text = "This is an amazing product. I love using it every day!"
```

```python
# Step 1: Tokenize
tokens = word_tokenize(text.lower())
```

```python
# Step 2: Remove stopwords & non-alphabetic tokens
stop_words = set(stopwords.words('english'))
filtered_tokens = [word for word in tokens if word not in stop_words and word.isalpha()]
```

```python
# Step 3: Apply stemming and lemmatizer

stemmed_tokens = [stemmer.stem(i) for i in filtered_tokens]
lemmatized_tokens = [lemmatizer.lemmatize(i, pos='v') for i in filtered_tokens]
```

```python
# Result
print("Original tokens:", tokens)
print("Filtered tokens:", filtered_tokens)
print("Stemmed tokens:", stemmed_tokens)
print("Lemmatized tokens:", lemmatized_tokens)
```
