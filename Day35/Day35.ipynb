# 🧠 Natural Language Processing (NLP) Overview

## 🤔 What is NLP?

**Definition:**  
Natural Language Processing (NLP) is a subfield of artificial intelligence (AI) focused on the interaction between computers and human (natural) languages. NLP enables machines to read, understand, and generate human language in a meaningful way.

---

## 🌍 Real-World Applications

- 💬 Chatbots like Siri, Alexa, and Google Assistant  
- 😊 Sentiment Analysis in social media or product reviews  
- 📰 Text Summarization for news or research articles  
- 🌐 Language Translation tools like Google Translate  

---

## 🗝️ Key NLP Tasks

- 🏷️ Text Classification (e.g., spam detection)
- 🏷️ Named Entity Recognition (NER) (e.g., extracting names, dates)
- 🏷️ Part-of-Speech Tagging (e.g., identifying nouns, verbs)
- 🌏 Machine Translation (e.g., converting languages)

---

# 📝 Text Data Representation

## 📊 Bag of Words (BoW)

- Represents text as an unordered collection of words.
- Each word is mapped to a feature, with its frequency as the value.

| Word      | I | love | NLP | machine | learning |
|-----------|---|------|-----|---------|----------|
| Text 1    | 1 | 1    | 1   | 0       | 0        |
| Text 2    | 1 | 1    | 0   | 1       | 1        |

## 📈 TF-IDF (Term Frequency-Inverse Document Frequency)

- Reflects the importance of a word in a document relative to all documents.
- Reduces the influence of common words and emphasizes important terms.

**Formula:**  
$TF\text{-}IDF = TF \times IDF$

# ✂️ Tokenization

**Definition:**  
Splitting text into individual words or tokens.

**Types:**
- Word Tokenization
- Sentence Tokenization

**Example:**
```python
sentence = "NLP is fun and interesting!"
tokens = word_tokenize(sentence)
print(tokens)
# Output: ['NLP', 'is', 'fun', 'and', 'interesting', '!']
```

---

# 🌱 Stemming and Lemmatization

**Stemming:**  
Reduces words to their root form (e.g., "Running" → "Run", "Happiness" → "Happi")

**Lemmatization:**  
Reduces words to their base form based on context (e.g., "Running" → "Run", "Better" → "Good")

**Difference:**  
- Stemming is faster but may produce non-dictionary words.
- Lemmatization is more accurate but computationally expensive.

**Example Code:**
```python
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
words = ['running', 'happiness', 'better']

stemmed_words = [stemmer.stem(word) for word in words]
print("Stemmed:", stemmed_words)
lemmatized_words = [lemmatizer.lemmatize(word, pos='v') for word in words]
print("Lemmatized:", lemmatized_words)
# Output:
# Stemmed: ['run', 'happi', 'better']
# Lemmatized: ['run', 'happiness', 'better']
```

# 🧪 Hands-on: Preprocessing Text Data

**Steps:**
1. Load a text dataset (e.g., product reviews)
2. Pre-process (tokenize, remove stop words, apply stemming/lemmatization)
3. Examine the pre-processed text

**Example Code:**
```python
text = "This is an amazing product. I love using it every day!"
tokens = word_tokenize(text.lower())
stop_words = set(stopwords.words('english'))
filtered_tokens = [word for word in tokens if word not in stop_words and word.isalpha()]
stemmer = PorterStemmer()
stemmed_tokens = [stemmer.stem(word) for word in filtered_tokens]
print(stemmed_tokens)
# Output: ['amaz', 'product', 'love', 'use', 'dai']
```

````python
# import modules
from nltk.tokenize import word_tokenize,sent_tokenize
from nltk.probability import FreqDist
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords
````

````python
# Sample Text
text = "NLP is fun. NLP is amazing."
````

````python
# Tokenization
tokens = sent_tokenize(text.lower())
print("Sentence Tokenize:",tokens)

tokens = word_tokenize(text.lower())
print("Word Tokenize:",tokens)
````

````python
# Calculate frequency
fdist = FreqDist(tokens)

print(fdist)
print(dict(fdist))

for word, freq in fdist.items():
    print(word, ":", freq)
````

````python
# Example words
words = ['running', 'happiness', 'better']
````

````python
# Stemming
stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(i) for i in words]
print('Stemmed:',stemmed_words)
````

````python
# Lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(i, pos='v') for i in words]
print('Lemmatized:',lemmatized_words)
````

````python
# Sample text
text = "This is an amazing product. I love using it every day!"
````

````python
# Step 1: Tokenize
tokens = word_tokenize(text.lower())
````

````python
# Step 2: Remove stopwords & non-alphabetic tokens
stop_words = set(stopwords.words('english'))
filtered_tokens = [word for word in tokens if word not in stop_words and word.isalpha()]
````

````python
# Step 3: Apply stemming and lemmatizer

stemmed_tokens = [stemmer.stem(i) for i in filtered_tokens]
lemmatized_tokens = [lemmatizer.lemmatize(i, pos='v') for i in filtered_tokens]
````

````python
# Result
print("Original tokens:", tokens)
print("Filtered tokens:", filtered_tokens)
print("Stemmed tokens:", stemmed_tokens)
print("Lemmatized tokens:", lemmatized_tokens)
````
